```python
import pandas as pd

path_to_df = "Data/combined_data_test.csv"
df = pd.read_csv(path_to_df)
df.head()
```

```python
#print the number of non rows with missing values in the "weekly_local_cases" column
num_missing = df["weekly_local_cases"].isna().sum()
print(f"Number of rows with missing values in 'weekly_local_cases': {num_missing}")

#print the percentage of complete rows in the dataframe with a value of 0 in the "weekly_local_cases" column
num_zero_cases = df[df["weekly_local_cases"] == 0].shape[0]
num_non_missing = df["weekly_local_cases"].notna().sum()
percentage_zero_cases = (num_zero_cases / num_non_missing) * 100
print(f"Percentage of non-missing rows with 0 weekly local cases: {percentage_zero_cases:.2f}%")
```

```python
#print a histogram of the distribution of values in the "weekly_local_cases" column
import matplotlib.pyplot as plt
plt.hist(df["weekly_local_cases"], bins=20)
plt.xlabel("Weekly Local Cases")
plt.ylabel("Frequency")
plt.title("Distribution of Weekly Local Cases")
plt.show()
```

```python
#make a graph showing the relationship between the "weekly_local_cases" column and the "Temp @ 60cm (C) Avg (Weekly Avg)" column
plt.scatter(df["Temp @ 60cm (C) Avg (Weekly Avg)"], df["weekly_local_cases"])
plt.xlabel("Temp @ 60cm (C) Avg (Weekly Avg)")
plt.ylabel("Weekly Local Cases")
plt.title("Relationship between Temperature and Weekly Local Cases")
plt.show()
```

```python
#test for the correlation between the "weekly_local_cases" column and the "Temp @ 60cm (C) Avg (Weekly Avg)" column
correlation = df["weekly_local_cases"].corr(df["Temp @ 60cm (C) Avg (Weekly Avg)"])
print(f"Correlation between weekly local cases and temperature: {correlation:.2f}")
```
