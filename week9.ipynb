## Introduction

Dengue virus transmission is tightly linked to climate and weather because its primary vectors, Aedes aegypti and Aedes albopictus, are highly sensitive to temperature, rainfall, and humidity. These climatic factors influence mosquito survival, breeding site availability, biting rates, and the speed of viral replication within the mosquito, which together determine the basic reproductive number and outbreak potential of dengue. Empirical and modeling studies show that dengue incidence exhibits clear seasonal and interannual patterns that track local climate variability and large-scale phenomena such as El Niño–Southern Oscillation, underscoring the climate sensitivity of transmission dynamics.

[Rocklöv, J](https://doi.org/10.1042/ETLS20180123) [Naish, S](https://doi.org/10.1186/1471-2334-14-167)

Temperature is often identified as the dominant climatic driver because it shapes multiple components of the transmission cycle simultaneously, including mosquito development time, adult longevity, blood-feeding frequency, and the extrinsic incubation period of dengue virus. Laboratory and meta-analytic work indicate that dengue transmission efficiency tends to peak within a moderate warm range (approximately $20–30 °C$), where vector competence and viral replication are optimized, while very low or very high temperatures suppress mosquito survival or virus development. Long-term analyses further demonstrate that warming trends can lengthen the seasonal window of transmission and expand the geographic and altitudinal range suitable for Aedes populations, shifting dengue risk into regions that were previously too cool for sustained transmission.

Rainfall and humidity modulate this temperature-dependent suitability by controlling the creation and persistence of aquatic habitats for immature mosquitoes and influencing adult mosquito survival and activity. Increased precipitation, especially following dry periods, typically generates new breeding sites in artificial containers and natural receptacles, while higher humidity is associated with enhanced Aedes feeding activity, survival, and egg development, thereby amplifying outbreak risk when susceptible human populations are present. However, the relationships are often nonlinear and lagged, with evidence that specific combinations and thresholds of temperature, rainfall, and relative humidity over preceding weeks to months are associated with subsequent increases in dengue incidence, highlighting the importance of integrated, climate-informed early warning and control strategies.

# Background

Dengue virus transmission is shaped by a complex interplay of viral, host, vector, and environmental factors, with climate and weather emerging as consistent predictors of temporal variation in incidence. Multiple studies demonstrate that temperature, humidity, and rainfall influence Aedes mosquito survival, development, biting behavior, and the duration of the extrinsic incubation period, thereby altering the effective reproductive number of dengue over short time scales. These relationships are often nonlinear and lagged, with warm, humid conditions and recent rainfall typically associated with increased transmission potential, while extreme heat or intense precipitation can reduce vector abundance or disrupt breeding sites. Much of the global evidence comes from endemic tropical regions, where dengue is hyperendemic and climate–transmission associations are evaluated using monthly or seasonal climate summaries, which may obscure finer-scale variability that is relevant for early warning in subtropical settings.

In the southeastern United States, and Florida in particular, climatic conditions already permit seasonal Aedes aegypti–mediated transmission, even though dengue is not uniformly endemic across the region. Modeling work suggests that large portions of Florida experience temperature and humidity profiles that support dengue transmission for extended parts of the year, especially in southern counties, where conditions in late summer and early fall are especially conducive. Observational analyses have begun to link interannual and intra-annual variation in local dengue case counts with fluctuations in temperature, relative humidity, and rainfall, indicating that relatively modest shifts in these parameters can change outbreak likelihood in this emerging-endemic context. However, prior studies often rely on coarser temporal resolutions or are restricted to specific counties or metropolitan areas, limiting inference about statewide climate–dengue relationships and their short-term dynamics.

Florida has reported recurrent clusters of locally acquired dengue since 2009, with notable increases in autochthonous transmission documented between 2022 and 2024 across multiple counties, particularly in South Florida. These events are occurring against a backdrop of rising temperatures and evolving precipitation patterns, heightening concern that ongoing climate change may expand the temporal window and geographic range of dengue risk in the state. In this context, a statewide dataset comprising weekly counts of locally acquired dengue cases alongside weekly averages of near-surface temperature at multiple heights, relative humidity, and cumulative rainfall from 2022 through early 2026 provides a valuable opportunity to characterize short-lag climate–transmission relationships at higher temporal resolution. This approach can help clarify which combinations and thresholds of temperature, humidity, and rainfall are most strongly associated with subsequent changes in dengue incidence in Florida, and can inform the development of climate-informed early warning systems and targeted seasonal preparedness efforts.
